```python
# TASK 2: Data Preparation & Integration
# Requirement: Load the customer, product and transaction datasets

import pandas as pd
import numpy as np
from pathlib import Path

# Set the recommendation data folder
data_folder = Path("../data/recommendation")

# Load the three CSV datasets
customers = pd.read_csv(data_folder / "Customers_repeat.csv")
products = pd.read_csv(data_folder / "Products_sort.csv")
transactions = pd.read_csv(data_folder / "Transactions_made.csv")

# Confirm that the datasets loaded successfully
print("Customers dataset loaded:", customers.shape)
print("Products dataset loaded:", products.shape)
print("Transactions dataset loaded:", transactions.shape)
```

```python
# TASK 2: Data Preparation & Integration
# Requirement: Inspect the structure and contents of the customer, product and transaction datasets

# Display the first five rows of each dataset
print("CUSTOMERS DATASET")
display(customers.head())

print("PRODUCTS DATASET")
display(products.head())

print("TRANSACTIONS DATASET")
display(transactions.head())
```

```python
# TASK 2: Data Preparation & Integration
# Requirement: Assess the quality of the customer, product and transaction datasets

# Check data quality for each dataset
datasets = {
    "Customers": customers,
    "Products": products,
    "Transactions": transactions
}

for name, df in datasets.items():
    print(f"\n{name} Dataset")
    print("-" * 30)
    print("Rows and columns:", df.shape)
    print("Missing values:")
    print(df.isnull().sum())
    print("Duplicate rows:", df.duplicated().sum())
    print("Data types:")
    print(df.dtypes)
```

```python
# TASK 2: Data Preparation & Integration
# Requirement: Assess potential duplicate transaction records

# Display the duplicated transaction records
duplicate_transactions = transactions[transactions.duplicated(keep=False)]

print("Number of duplicate transaction rows:", len(duplicate_transactions))

display(duplicate_transactions.head(20))
```

```python
# TASK 2: Data Preparation & Integration
# Requirement: Clean the transaction dataset by removing exact duplicate records

# Record the original number of transaction rows
original_rows = len(transactions)

# Remove exact duplicate rows
transactions = transactions.drop_duplicates()

# Calculate how many rows were removed
removed_rows = original_rows - len(transactions)

# Display the cleaning results
print("Original transaction rows:", original_rows)
print("Duplicate rows removed:", removed_rows)
print("Remaining transaction rows:", len(transactions))
print("Remaining duplicates:", transactions.duplicated().sum())
```

```python
# TASK 2: Data Preparation & Integration
# Requirement: Investigate missing product ratings as part of data quality assessment

# Display descriptive statistics for product ratings
print("Product rating statistics:")
display(products["Rating"].describe())

# Count missing and available ratings
print("Missing ratings:", products["Rating"].isnull().sum())
print("Available ratings:", products["Rating"].notnull().sum())
```

```python
# TASK 2: Data Preparation & Integration
# Requirement: Validate customer and product identifiers before dataset integration

# Check whether transaction customer IDs exist in the Customers dataset
unmatched_customers = transactions[
    ~transactions["CustomerID"].isin(customers["CustomerID"])
]

# Check whether transaction product IDs exist in the Products dataset
unmatched_products = transactions[
    ~transactions["ProductID"].isin(products["ProductID"])
]

# Display the results
print("Transactions with unmatched customer IDs:", len(unmatched_customers))
print("Transactions with unmatched product IDs:", len(unmatched_products))
```

```python
# TASK 2: Data Preparation & Integration
# Requirement: Integrate customer, transaction and product datasets

# Merge transactions with customer information
integrated_data = transactions.merge(
    customers,
    on="CustomerID",
    how="left"
)

# Merge the result with product information
integrated_data = integrated_data.merge(
    products,
    on="ProductID",
    how="left"
)

# Display the integrated dataset dimensions
print("Integrated dataset dimensions:", integrated_data.shape)

# Display the first five records
display(integrated_data.head())
```

```python
# TASK 2: Data Preparation & Integration
# Requirement: Create a customer-product interaction matrix

# Create a matrix showing purchase counts for each customer and product
interaction_matrix = integrated_data.pivot_table(
    index="CustomerID",
    columns="ProductID",
    values="PurchaseCount",
    aggfunc="sum",
    fill_value=0
)

# Display the interaction matrix dimensions
print("Interaction matrix dimensions:", interaction_matrix.shape)

# Display the first five customers and their product interactions
display(interaction_matrix.head())
```

```python
# TASK 2: Data Preparation & Integration
# Requirement: Verify customer-product interactions in the interaction matrix

# Count the number of customer-product combinations with purchases
nonzero_interactions = (interaction_matrix > 0).sum().sum()

# Calculate the percentage of combinations with purchases
total_combinations = interaction_matrix.size
interaction_percentage = (nonzero_interactions / total_combinations) * 100

print("Total customer-product combinations:", total_combinations)
print("Combinations with purchases:", nonzero_interactions)
print("Interaction density:", round(interaction_percentage, 2), "%")

# Display Customer 1's recorded product purchases
print("\nCustomer 1's product interactions:")
display(interaction_matrix.loc[1][interaction_matrix.loc[1] > 0])
```

```python

# TASK 2: Customer Analytics
# Requirement: Identify the top 10 customers by total spend

customer_spend = transactions.groupby("CustomerID")["Spend"].sum()

top_10_spend = customer_spend.sort_values(ascending=False).head(10)

print("Top 10 Customers by Total Spend:")
display(top_10_spend)
```

```python

# TASK 2: Customer Analytics
# Requirement: Identify the top 10 customers by purchase frequency

customer_frequency = transactions.groupby("CustomerID")["PurchaseCount"].sum()

top_10_frequency = customer_frequency.sort_values(ascending=False).head(10)

print("Top 10 Customers by Purchase Frequency:")
display(top_10_frequency)
```

```python

# TASK 2: Customer Analytics
# Requirement: Analyse customer value using spending and purchase frequency

customer_value = transactions.groupby("CustomerID").agg(
    TotalSpend=("Spend", "sum"),
    TotalPurchases=("PurchaseCount", "sum")
)

customer_value["AverageSpendPerPurchase"] = (
    customer_value["TotalSpend"] / customer_value["TotalPurchases"]
)

print("Customer Value Analysis:")
display(customer_value.head(10))
```

```python

# TASK 2: Customer Analytics
# Requirement: Compare the spending and purchasing behaviour of the highest-value customers

top_10_customer_value = customer_value.sort_values(
    by="TotalSpend",
    ascending=False
).head(10)

print("Top 10 Customers - Customer Value Comparison:")
display(top_10_customer_value)
```

```python

# TASK 2: Product Analytics
# Requirement: Identify the top 10 products by total revenue

product_revenue = transactions.groupby("ProductID")["Spend"].sum()

top_10_products = product_revenue.sort_values(
    ascending=False
).head(10)

print("Top 10 Products by Revenue:")
display(top_10_products)
```

```python

# TASK 2: Product Analytics
# Requirement: Analyse total revenue generated by each product category

category_revenue = integrated_data.groupby("Category")["Spend"].sum()

category_revenue = category_revenue.sort_values(ascending=False)

print("Revenue by Product Category:")
display(category_revenue)
```

```python

# TASK 2: Product Analytics
# Requirement: Analyse total revenue generated by each product brand

brand_revenue = integrated_data.groupby("Brand")["Spend"].sum()

brand_revenue = brand_revenue.sort_values(ascending=False)

print("Revenue by Product Brand:")
display(brand_revenue)
```

```python

# TASK 2: GroupBy & Pivot Table Analysis
# Requirement: Analyse customer spending and purchasing behaviour by region

region_analysis = integrated_data.groupby("Region").agg(
    TotalRevenue=("Spend", "sum"),
    TotalPurchases=("PurchaseCount", "sum"),
    NumberOfCustomers=("CustomerID", "nunique")
)

region_analysis["AverageSpendPerCustomer"] = (
    region_analysis["TotalRevenue"] / region_analysis["NumberOfCustomers"]
)

region_analysis = region_analysis.sort_values(
    by="TotalRevenue",
    ascending=False
)

print("Customer Performance by Region:")
display(region_analysis)
```

```python

# TASK 2: GroupBy & Pivot Table Analysis
# Requirement: Analyse revenue and purchasing behaviour by customer loyalty level

loyalty_analysis = integrated_data.groupby("LoyaltyLevel").agg(
    TotalRevenue=("Spend", "sum"),
    TotalPurchases=("PurchaseCount", "sum"),
    NumberOfCustomers=("CustomerID", "nunique")
)

loyalty_analysis["AverageSpendPerCustomer"] = (
    loyalty_analysis["TotalRevenue"] / loyalty_analysis["NumberOfCustomers"]
)

loyalty_analysis = loyalty_analysis.sort_values(
    by="TotalRevenue",
    ascending=False
)

print("Customer Performance by Loyalty Level:")
display(loyalty_analysis)
```

```python

# TASK 2: GroupBy & Pivot Table Analysis
# Requirement: Compare product category revenue and purchasing performance

category_analysis = integrated_data.groupby("Category").agg(
    TotalRevenue=("Spend", "sum"),
    TotalPurchases=("PurchaseCount", "sum"),
    NumberOfProducts=("ProductID", "nunique")
)

category_analysis = category_analysis.sort_values(
    by="TotalRevenue",
    ascending=False
)

print("Product Category Performance:")
display(category_analysis)
```

```python

# TASK 2: GroupBy & Pivot Table Analysis
# Requirement: Create a pivot table comparing revenue by region and customer loyalty level

region_loyalty_pivot = pd.pivot_table(
    integrated_data,
    index="Region",
    columns="LoyaltyLevel",
    values="Spend",
    aggfunc="sum",
    fill_value=0
)

print("Revenue by Region and Loyalty Level:")
display(region_loyalty_pivot)
```

```python

# TASK 2: GroupBy & Pivot Table Analysis
# Requirement: Create a pivot table comparing revenue by product category and brand

category_brand_pivot = pd.pivot_table(
    integrated_data,
    index="Category",
    columns="Brand",
    values="Spend",
    aggfunc="sum",
    fill_value=0
)

print("Revenue by Product Category and Brand:")
display(category_brand_pivot)
```

```python
# TASK 2: Recommendation System Development
# Requirement: Develop a popularity-based recommendation system

# Calculate total purchases for each product
product_popularity = interaction_matrix.sum(axis=0)

# Rank products from most to least popular
popular_products = product_popularity.sort_values(ascending=False)

# Display the 10 most popular products
print("Top 10 Most Popular Products:")
display(popular_products.head(10))
```

```python
# TASK 2: Recommendation System Development
# Requirement: Display popular products with their product details

# Convert the popularity ranking into a DataFrame
popular_products_df = popular_products.reset_index(name="TotalPurchases")

# Add product details using ProductID
popular_products_df = popular_products_df.merge(
    products,
    on="ProductID",
    how="left"
)

# Display the 10 most popular products
print("Top 10 Most Popular Products with Details:")
display(popular_products_df.head(10))
```

```python
# TASK 2: Recommendation System Development
# Requirement: Create a function to recommend popular products to customers

def recommend_popular_products(customer_id, number_of_recommendations=5):

    # Identify products already purchased by this customer
    customer_purchases = interaction_matrix.loc[customer_id]
    purchased_products = customer_purchases[customer_purchases > 0].index

    # Exclude products the customer has already purchased
    recommendations = popular_products[
        ~popular_products.index.isin(purchased_products)
    ]

    # Select the requested number of recommendations
    recommendations = recommendations.head(number_of_recommendations)

    # Add product details
    recommendations = recommendations.reset_index(name="TotalPurchases")
    recommendations = recommendations.merge(
        products,
        on="ProductID",
        how="left"
    )

    return recommendations

print("Popularity-based recommendation function created successfully.")
```

```python
# TASK 2: Recommendation System Development
# Requirement: Test the popularity-based recommendation system

customer_id = 1

recommendations = recommend_popular_products(customer_id)

print(f"Popular product recommendations for Customer {customer_id}:")
display(recommendations)
```

```python
# TASK 2: Recommendation System Development
# Requirement: Verify that previously purchased products are excluded

# Find the most popular product
most_popular_product = popular_products.index[0]

# Find a customer who has already purchased it
test_customer = interaction_matrix.index[
    interaction_matrix[most_popular_product] > 0
][0]

# Generate recommendations for that customer
test_recommendations = recommend_popular_products(test_customer)

print("Test customer:", test_customer)
print("Previously purchased product:", most_popular_product)
print("Is that product excluded?",
      most_popular_product not in test_recommendations["ProductID"].values)

display(test_recommendations)
```

```python
# TASK 2: Recommendation System Development
# Requirement: Calculate customer similarities for collaborative filtering

from sklearn.metrics.pairwise import cosine_similarity

# Calculate similarity between every pair of customers
customer_similarity = cosine_similarity(interaction_matrix)

# Convert the similarity results into a DataFrame
customer_similarity_df = pd.DataFrame(
    customer_similarity,
    index=interaction_matrix.index,
    columns=interaction_matrix.index
)

print("Customer similarity matrix dimensions:", customer_similarity_df.shape)
display(customer_similarity_df.iloc[:5, :5])
```

```python
# TASK 2: Recommendation System Development
# Requirement: Identify customers with similar purchasing patterns

customer_id = 4

# Retrieve similarity scores for the selected customer
similar_customers = customer_similarity_df.loc[customer_id]

# Exclude the customer themselves
similar_customers = similar_customers.drop(customer_id)

# Rank other customers by similarity
top_similar_customers = similar_customers.sort_values(
    ascending=False
).head(5)

print(f"Five most similar customers to Customer {customer_id}:")
display(top_similar_customers)
```

```python
# TASK 2: Recommendation System Development
# Requirement: Examine the purchasing behaviour of similar customers

# Get the purchase records for the five most similar customers
similar_customer_purchases = integrated_data[
    integrated_data["CustomerID"].isin(top_similar_customers.index)
]

print("Products purchased by the five most similar customers:")
display(
    similar_customer_purchases[
        ["CustomerID", "ProductID", "PurchaseCount", "Category", "Brand"]
    ].sort_values("CustomerID")
)
```

```python
# TASK 2: Recommendation System Development
# Requirement: Identify a broader group of similar customers

customer_id = 4

# Retrieve similarity scores for Customer 4
similar_customers = customer_similarity_df.loc[customer_id]

# Exclude Customer 4
similar_customers = similar_customers.drop(customer_id)

# Identify the 20 most similar customers
top_similar_customers = similar_customers.sort_values(
    ascending=False
).head(20)

print("20 most similar customers to Customer 4:")
display(top_similar_customers)
```

```python
# TASK 2: Recommendation System Development
# Requirement: Examine products purchased by the 20 most similar customers

similar_customer_purchases = integrated_data[
    integrated_data["CustomerID"].isin(top_similar_customers.index)
]

print("Purchases made by the 20 most similar customers:")

display(
    similar_customer_purchases[
        ["CustomerID", "ProductID", "PurchaseCount", "Category", "Brand"]
    ].sort_values(["CustomerID", "ProductID"])
)
```

```python
# TASK 2: Recommendation System Development
# Requirement: Calculate product recommendation scores using collaborative filtering

# Retrieve purchase histories of the 20 most similar customers
similar_customer_matrix = interaction_matrix.loc[top_similar_customers.index]

# Weight each customer's purchases by their similarity to Customer 4
recommendation_scores = similar_customer_matrix.T.dot(top_similar_customers)

# Exclude products Customer 4 has already purchased
customer_purchases = interaction_matrix.loc[customer_id]
recommendation_scores = recommendation_scores[
    customer_purchases == 0
]

# Rank products by recommendation score
recommendation_scores = recommendation_scores.sort_values(
    ascending=False
)

print("Top 10 Collaborative Filtering Recommendations:")
display(recommendation_scores.head(10))
```

```python
# TASK 2: Recommendation System Development
# Requirement: Display collaborative filtering recommendations with product details

# Convert recommendation scores into a DataFrame
collaborative_recommendations = recommendation_scores.reset_index(
    name="RecommendationScore"
)

# Add product information
collaborative_recommendations = collaborative_recommendations.merge(
    products,
    on="ProductID",
    how="left"
)

# Display the top 10 recommendations
print("Collaborative Filtering Recommendations for Customer 4:")
display(collaborative_recommendations.head(10))
```

```python
# TASK 2: Recommendation System Development
# Requirement: Create a collaborative filtering recommendation function

def recommend_collaborative_products(customer_id, number_of_recommendations=5):

    # Find the 20 most similar customers
    similarities = customer_similarity_df.loc[customer_id].drop(customer_id)
    similar_customers = similarities.sort_values(ascending=False).head(20)

    # Calculate weighted product recommendation scores
    similar_purchases = interaction_matrix.loc[similar_customers.index]
    scores = similar_purchases.T.dot(similar_customers)

    # Exclude products already purchased by the customer
    scores = scores[interaction_matrix.loc[customer_id] == 0]

    # Select the highest-scoring products
    scores = scores.sort_values(ascending=False).head(number_of_recommendations)

    # Add product details
    recommendations = scores.reset_index(name="RecommendationScore")
    recommendations = recommendations.merge(products, on="ProductID", how="left")

    return recommendations

print("Collaborative filtering recommendation function created successfully.")
```

```python
# TASK 2: Recommendation System Development
# Requirement: Test the collaborative filtering recommendation system

recommendations = recommend_collaborative_products(4)

print("Collaborative filtering recommendations for Customer 4:")
display(recommendations)
```

```python
# TASK 2: Recommendation Evaluation
# Requirement: Generate recommendations for at least five example customers

example_customers = [1, 2, 3, 4, 5]

for customer_id in example_customers:

    print(f"\nCUSTOMER {customer_id}")
    print("-" * 50)

    print("Popularity-based recommendations:")
    display(recommend_popular_products(customer_id, 5))

    print("Collaborative filtering recommendations:")
    display(recommend_collaborative_products(customer_id, 5))
```

```python
# TASK 2: Recommendation Evaluation
# Requirement: Compare the outputs of both recommendation approaches

comparison_results = []

for customer_id in [1, 2, 3, 4, 5]:

    popular = recommend_popular_products(customer_id, 5)
    collaborative = recommend_collaborative_products(customer_id, 5)

    popular_ids = set(popular["ProductID"])
    collaborative_ids = set(collaborative["ProductID"])

    common_products = popular_ids.intersection(collaborative_ids)

    comparison_results.append({
        "CustomerID": customer_id,
        "PopularityRecommendations": len(popular_ids),
        "CollaborativeRecommendations": len(collaborative_ids),
        "CommonProducts": len(common_products),
        "DifferentProducts": len(popular_ids.symmetric_difference(collaborative_ids))
    })

comparison_df = pd.DataFrame(comparison_results)

print("Recommendation Approach Comparison:")
display(comparison_df)
```

```python
# TASK 2: Recommendation Evaluation
# Requirement: Calculate the average overlap between recommendation approaches

average_overlap = comparison_df["CommonProducts"].mean()

average_overlap_percentage = (average_overlap / 5) * 100

print("Average number of common recommendations:",
      round(average_overlap, 2))

print("Average recommendation overlap:",
      round(average_overlap_percentage, 2), "%")
```

```python
# TASK 2: Recommendation Evaluation
# Requirement: Compare the strengths and limitations of both recommendation approaches

evaluation = pd.DataFrame({
    "EvaluationCriteria": [
        "Recommendation basis",
        "Personalisation",
        "Ease of implementation",
        "Main strength",
        "Main limitation"
    ],
    "PopularityBased": [
        "Overall product purchase frequency",
        "Low",
        "Simple",
        "Identifies broadly popular products",
        "Does not account for individual preferences"
    ],
    "CollaborativeFiltering": [
        "Purchasing patterns of similar customers",
        "Higher",
        "More complex",
        "Provides customer-specific recommendations",
        "Affected by sparse purchasing data"
    ]
})

print("Recommendation System Evaluation:")
display(evaluation)
```

Recommendation Evaluation

The popularity-based recommendation system provides a straightforward approach by ranking products according to their total purchase frequency. It is useful for identifying products with broad customer appeal, but does not account for individual purchasing preferences.

Collaborative filtering uses cosine similarity to identify customers with similar purchasing patterns and generates recommendations based on their purchase histories. This provides a more personalised approach, although its effectiveness can be affected by sparse customer-product interactions.

When comparing the five example customers, both systems generated five recommendations each. The average overlap between their recommendations was only 12%, demonstrating that the approaches frequently suggested different products.

Collaborative filtering is the stronger approach when the objective is to provide personalised recommendations. However, popularity-based filtering remains a useful baseline because it is simpler and identifies products with broad appeal. Neither approach has been proven to achieve higher purchase conversion or predictive accuracy, as these were not directly tested.
